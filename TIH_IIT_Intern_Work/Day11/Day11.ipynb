```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/drive/MyDrive/TIH_IIT_Intern_Work/Dataset.csv")

df.head()
```

## 1. Drop Unnecessary Column

```python
df = df.drop(
    columns=["S No"]
)
```

## 2. Encode NTU

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df["NTU"] = encoder.fit_transform(
    df["NTU"]
)

df.head()
```

## 3. Define Features & Target

```python
X = df[
    [
        "Analog value",
        "Analog Voltage",
        "NTU"
    ]
]

y = df["Binary bit S"]
```

## 4. Train-Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
```

## 5. Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(
    X_train
)

X_test = scaler.transform(
    X_test
)
```

## 6. Preprocessed Data

```python
x_train = pd.DataFrame(X_train, columns=["Analog value", "Analog Voltage", "NTU"])
x_test = pd.DataFrame(X_test, columns=["Analog value", "Analog Voltage", "NTU"])

x_train.head()
```

```python
x_train.to_csv('x_train_preprocessed.csv', index=False)
print('x_train_preprocessed.csv has been saved. You can download it from the file explorer on the left.')
```

```python
x_test.to_csv('x_test_preprocessed.csv', index=False)
print('x_test_preprocessed.csv has been saved. You can download it from the file explorer on the left.')
```
