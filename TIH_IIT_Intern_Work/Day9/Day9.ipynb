## 4. Invalid Bits

```python
for col in ["binary bit R", "Binary bit S"]:
    if col in df.columns:
        invalid = df[~df[col].isin([0,1])]
        print(f"\nInvalid values in {col}: {len(invalid)}")
```

## 5. Class Imbalance

```python
print("\nCLASS BALANCE")
print(df["Binary bit S"].value_counts())
print("\nPercentage")
print(round(df["Binary bit S"].value_counts(normalize=True)*100,2))
```

## 6. Outlier Detection

```python
Q1 = df['Analog Voltage'].quantile(0.25)
Q3 = df['Analog Voltage'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Analog Voltage'] < lower_bound) | (df['Analog Voltage'] > upper_bound)]

print(f"Outliers in 'Analog Voltage' column (using IQR method):")
display(outliers)
print(f"Total number of outliers: {len(outliers)}")
```
